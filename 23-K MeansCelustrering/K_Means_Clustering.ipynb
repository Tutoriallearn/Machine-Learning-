```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
```

```python
X = np.array([
    [1,2],
    [1.5,1.8],
    [5,8],
    [8,8],
    [1,0.6],
    [9,11],
    [8,2],
    [10,2],
    [9,3]
])
```

```python

wcss = []

# IMPORTANT FIX
for i in range(1, len(X)+1):

    kmeans = KMeans(
        n_clusters=i,
        init='k-means++',
        random_state=42,
        n_init=10
    )

    kmeans.fit(X)

    wcss.append(kmeans.inertia_)

# Plot
plt.plot(range(1, len(X)+1), wcss, marker='o')

plt.title("Elbow Method")
plt.xlabel("Number of Clusters")
plt.ylabel("WCSS")

plt.show()
```

```python
kmeans = KMeans(n_clusters=3)

kmeans.fit(X)
```

```python
labels = kmeans.labels_
centroids = kmeans.cluster_centers_

print("Cluster Labels:")
print(labels)

print("\nCentroids:")
print(centroids)
```

```python
plt.scatter(X[:,0], X[:,1], c=labels, cmap='rainbow')

plt.scatter(
    centroids[:,0],
    centroids[:,1],
    color='black',
    marker='X',
    s=200
)

plt.title("K-Means Clustering")
plt.show()
```
