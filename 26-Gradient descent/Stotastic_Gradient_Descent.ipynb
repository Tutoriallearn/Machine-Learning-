```python
import numpy as np
import random
```

```python
# Dataset
X = np.array([1, 2, 3, 4], dtype=float)
Y = np.array([2, 4, 6, 8], dtype=float)
```

```python
# Initialize parameters
m = 0
b = 0
```

```python
# Hyperparameters
learning_rate = 0.01
epochs = 100

n = len(X)
```

```python
# SGD Training
for epoch in range(epochs):

    # Shuffle data
    indices = list(range(n))
    random.shuffle(indices)

    for i in indices:

        x = X[i]
        y = Y[i]

        # Prediction
        y_pred = m * x + b

        # Gradients
        dm = -2 * x * (y - y_pred)
        db = -2 * (y - y_pred)

        # Update parameters
        m = m - learning_rate * dm
        b = b - learning_rate * db

    # Loss calculation
    total_pred = m * X + b
    loss = np.mean((Y - total_pred) ** 2)

    if epoch % 10 == 0:
        print(f"Epoch {epoch}, Loss: {loss:.4f}")
```

```python
print("\nFinal Parameters:")
print("m =", m)
print("b =", b)
```
