```python
import numpy as np
```

```python
# Input: 5 samples, 2 features
X = np.array([
    [2.0, 1.0],
    [1.5, 2.0],
    [3.0, 1.5],
    [3.5, 3.0],
    [2.5, 2.0]
])

# Labels (0 or 1)
y = np.array([0, 0, 1, 1, 1])
```

```python
#Sigmoid (activation) function.
def sigmoid(z):

    return 1 / (1 + np.exp(-z))
```

```python
#Compute probability P(y=1|x).
def predict_proba(X, w, b):
    z = np.dot(X, w) + b
    return sigmoid(z)
```

```python
 #Predict class labels (0 or 1).
def predict(X, w, b, threshold=0.5):
    probs = predict_proba(X, w, b)
    return (probs >= threshold).astype(int)
```

```python
def compute_loss(X, y, w, b):
    #Binary cross-entropy loss.
    m = X.shape[0]
    probs = predict_proba(X, w, b)

    # Avoid log(0)
    probs = np.clip(probs, 1e-15, 1 - 1e-15)

    loss = -1/m * np.sum(
        y * np.log(probs) +
        (1 - y) * np.log(1 - probs)
    )

    return loss
```

```python
#Train Logistic Regression using Gradient Descent.
def train_logistic_regression(X, y, lr=0.1, num_iters=1000):


    m, n = X.shape
    w = np.zeros(n)
    b = 0.0

    for i in range(num_iters):

        probs = predict_proba(X, w, b)

        dz = probs - y
        dw = (1/m) * np.dot(X.T, dz)
        db = (1/m) * np.sum(dz)

        w -= lr * dw
        b -= lr * db

        if i % 100 == 0 or i == num_iters - 1:
            loss = compute_loss(X, y, w, b)
            print(f"Iter {i:4d}, Loss: {loss:.6f}")

    return w, b
```

```python
# Train model
w, b = train_logistic_regression(X, y, lr=0.5, num_iters=1000)

```

```python
# Predictions
probs = predict_proba(X, w, b)
preds = predict(X, w, b, threshold=0.5)

```

```python
print("Probabilities:", probs)
print("Predicted Classes:", preds)
print("Final Weights:", w)
print("Final Bias:", b)
```
