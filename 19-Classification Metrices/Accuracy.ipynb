```python
#Define Data Set
y_actual = [1, 0, 1, 1, 0]
y_pred = [1, 0, 0, 1, 1]
```

```python
#Intialise as 0
TP = TN = FP = FN = 0
```

```python
# loop to entire y-actual dataset
for i in range(len(y_actual)):
    if y_actual[i] == 1 and y_pred[i] == 1:
        TP += 1
    elif y_actual[i] == 0 and y_pred[i] == 0:
        TN += 1
    elif y_actual[i] == 0 and y_pred[i] == 1:
        FP += 1
    elif y_actual[i] == 1 and y_pred[i] == 0:
        FN += 1
```

```python
#Apply Accuracy Formula
accuracy = (TP + TN) / (TP + TN + FP + FN)
```

```python
#Find Value
accuracy
```
