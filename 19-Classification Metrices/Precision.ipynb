```python
#Define Data Set
y_actual = [1, 0, 1, 1, 0]
y_pred = [1, 0, 0, 1, 1]

```

```python
#Intialise as 0
TP = FP = 0

```

```python
# loop to entire y-actual dataset
for i in range(len(y_actual)):
    if y_pred[i] == 1:
        if y_actual[i] == 1:
            TP += 1
        else:
            FP += 1

```

```python
#Apply Precision Formula
precision = TP / (TP + FP)

```

```python
#Find Value
precision
```

```python
from sklearn.metrics import precision_score

y_actual = [1, 0, 1, 1, 0]
y_pred = [1, 0, 0, 1, 1]

```

```python
precision = precision_score(y_actual, y_pred)
```
