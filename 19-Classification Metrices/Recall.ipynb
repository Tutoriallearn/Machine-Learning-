```python
#Define Data Set
y_actual = [1, 0, 1, 1, 0]
y_pred = [1, 0, 0, 1, 1]

```

```python
#Intialise as 0
TP = FN = 0

```

```python
# loop to entire y-actual dataset
for i in range(len(y_actual)):
    if y_actual[i] == 1:
        if y_pred[i] == 1:
            TP += 1
        else:
            FN += 1

```

```python
#Apply Recall Formula
recall = TP / (TP + FN)

```

```python
#Find Value
recall
```
